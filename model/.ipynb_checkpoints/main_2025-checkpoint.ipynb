```python
from sklearn.svm import SVC
from tensorflow import keras
import matplotlib.pyplot as plt

class MakeModel:
    # 1st try without neural nets        
    def svc(c: float = 0.1):
        svc = SVC(kernel='rbf',
                  C=c,
                  gamma='scale',
                  random_state=42)
        return svc
    
    # simple keras model
    def keras_sequential(input_shape, num_classes):  # Fixed spelling and parameters
        model = keras.Sequential([  # Fixed spelling and added list
            keras.layers.Dense(512, activation='relu', input_shape=input_shape),  # Fixed Layers to layers
            keras.layers.BatchNormalization(),
            keras.layers.Dropout(0.25),

            keras.layers.Dense(256, activation='relu'),
            keras.layers.BatchNormalization(),
            keras.layers.Dropout(0.25),

            keras.layers.Dense(128, activation='relu'),
            keras.layers.BatchNormalization(),
            keras.layers.Dropout(0.3),

            keras.layers.Dense(num_classes, activation='softmax')
        ])

        model.compile(
            optimizer='adam',
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy']
        )

        return model

class Visualize:  # author: Deepseek
    
    def plot_training_history(history):
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
        
        # Accuracy
        ax1.plot(history.history['accuracy'], label='Training Accuracy')
        ax1.plot(history.history['val_accuracy'], label='Validation Accuracy')
        ax1.set_title('Model Accuracy')
        ax1.set_xlabel('Epoch')
        ax1.set_ylabel('Accuracy')
        ax1.legend()
        
        # Loss
        ax2.plot(history.history['loss'], label='Training Loss')
        ax2.plot(history.history['val_loss'], label='Validation Loss')
        ax2.set_title('Model Loss')
        ax2.set_xlabel('Epoch')
        ax2.set_ylabel('Loss')
        ax2.legend()
        
        plt.tight_layout()
        plt.show()
```
